```python
# Outlier analysis using the IQR method

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

features = [
    "pH", "EC", "TDS", "TH", "Ca", "Mg", "Na", "K",
    "Cl", "SO4", "HCO3", "NO3", "F",
    "As", "Pb", "Cd", "Cr", "Ni", "Cu", "Zn", "Fe", "Mn"
]

outlier_results = []

for feature in features:
    Q1 = df[feature].quantile(0.25)
    Q3 = df[feature].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = (
        (df[feature] < lower_bound) |
        (df[feature] > upper_bound)
    )

    outlier_results.append({
        "Feature": feature,
        "Lower_Bound": lower_bound,
        "Upper_Bound": upper_bound,
        "Outlier_Count": int(outliers.sum()),
        "Outlier_Percentage": round(
            outliers.mean() * 100, 2
        )
    })

outlier_summary = pd.DataFrame(outlier_results)

display(
    outlier_summary.sort_values(
        "Outlier_Percentage",
        ascending=False
    ).round(3)
)
```

```python

from sklearn.model_selection import StratifiedKFold, cross_val_predict
from sklearn.metrics import (classification_report, confusion_matrix,
                             average_precision_score, matthews_corrcoef)

X = df[features]
y = 1 - df["Is_Unsafe"]            # 1 = Safe (rare class), 0 = Unsafe

n_safe, n_unsafe = (y == 1).sum(), (y == 0).sum()
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

def xgb(**kw):
    return XGBClassifier(n_estimators=200, max_depth=3, learning_rate=0.05,
                         eval_metric="logloss", random_state=42, **kw)

models = {
    "BalancedRF":   BalancedRandomForestClassifier(n_estimators=300, max_depth=6, random_state=42),
    "XGB_weighted": xgb(scale_pos_weight=n_unsafe / n_safe),
    "XGB_SMOTE":    Pipeline([("smote", SMOTE(sampling_strategy=0.3, k_neighbors=5, random_state=42)),
                              ("clf", xgb())]),
}

summary = {}
for name, m in models.items():
    proba = cross_val_predict(m, X, y, cv=cv, method="predict_proba", n_jobs=-1)[:, 1]
    pred = (proba >= 0.5).astype(int)

    print("=" * 60)
    print(name)
    print("=" * 60)
    print(classification_report(y, pred, target_names=["Unsafe (0)", "Safe (1)"], digits=3))
    cm = confusion_matrix(y, pred)
    print(pd.DataFrame(cm, index=["Actual Unsafe", "Actual Safe"],
                       columns=["Pred Unsafe", "Pred Safe"]))
    print()

    summary[name] = {
        "safe_precision": cm[1, 1] / max(cm[:, 1].sum(), 1),
        "safe_recall":    cm[1, 1] / cm[1].sum(),
        "pr_auc":         average_precision_score(y, proba),
        "mcc":            matthews_corrcoef(y, pred),
        "false_safe":     int(cm[0, 1]),
    }

print(pd.DataFrame(summary).T.round(3))
```

```python

from sklearn.ensemble import VotingClassifier
from sklearn.model_selection import StratifiedKFold, cross_val_predict
from sklearn.metrics import average_precision_score, matthews_corrcoef

X = df[features]
y = 1 - df["Is_Unsafe"]                     # 1 = Safe
n_safe, n_unsafe = (y == 1).sum(), (y == 0).sum()
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

def xgb(depth=3):
    return XGBClassifier(n_estimators=300, max_depth=depth, learning_rate=0.05,
                         scale_pos_weight=n_unsafe / n_safe,
                         eval_metric="logloss", random_state=42)

brf = BalancedRandomForestClassifier(n_estimators=300, max_depth=6, random_state=42)

candidates = {
    "XGB depth3": xgb(3),
    "XGB depth6": xgb(6),
    "BalancedRF": brf,
    "Voting":     VotingClassifier([("x", xgb(3)), ("b", brf)], voting="soft"),
}

rows = []
for name, m in candidates.items():
    p = cross_val_predict(m, X, y, cv=cv, method="predict_proba", n_jobs=-1)[:, 1]
    pr = average_precision_score(y, p)
    for t in [0.5, 0.7, 0.8, 0.9, 0.95]:
        pred = p >= t
        tp = (pred & (y == 1)).sum()
        fp = (pred & (y == 0)).sum()
        rows.append({"model": name, "PR-AUC": round(pr, 3), "threshold": t,
                     "safe_found": f"{tp}/{n_safe}", "false_safe": fp,
                     "safe_precision": round(tp / max(tp + fp, 1), 3),
                     "MCC": round(matthews_corrcoef(y, pred), 3)})
print(pd.DataFrame(rows).to_string(index=False))
```

```python


ops = {"XGB d3 @0.9": (xgb(3), 0.9),
       "XGB d6 @0.8": (xgb(6), 0.8),
       "BRF @0.8":    (brf, 0.8),
       "BRF @0.9":    (brf, 0.9)}

rows = []
for name, (m, t) in ops.items():
    found, fs = [], []
    for seed in range(5):
        cv_s = StratifiedKFold(n_splits=5, shuffle=True, random_state=seed)
        p = cross_val_predict(m, X, y, cv=cv_s, method="predict_proba", n_jobs=-1)[:, 1]
        pred = p >= t
        found.append((pred & (y == 1)).sum())
        fs.append((pred & (y == 0)).sum())
    rows.append({"operating point": name,
                 "safe found (mean ± sd)": f"{np.mean(found):.1f} ± {np.std(found):.1f}",
                 "false-safe (mean ± sd)": f"{np.mean(fs):.1f} ± {np.std(fs):.1f}",
                 "false-safe range": f"{min(fs)}-{max(fs)}"})
print(pd.DataFrame(rows).to_string(index=False))
```
