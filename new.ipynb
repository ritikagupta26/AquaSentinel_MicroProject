```python
import pandas as pd

raw = pd.read_csv("dataset/delhi_20yr.csv")

keep = ["TDS", "TH", "F", "NO3", "K",
        "As", "Pb", "Cd", "Cr", "Ni", "Cu", "Zn", "Fe", "Mn", "pH"]

slim = raw[["Year", "Latitude", "Longitude"] + keep + ["Is_Unsafe"]].copy()

# years become columns, elements become rows
year_table = slim.groupby("Year")[keep].mean().T

# extra row: share of unsafe samples in each year
year_table.loc["Unsafe_share"] = slim.groupby("Year")["Is_Unsafe"].mean()

pd.set_option("display.width", 300)
pd.set_option("display.max_columns", 30)
print(year_table.shape)        # expect (16, 21): 15 elements + Unsafe_share, 21 years
year_table.round(3)
```

```python
# rows = (Location, element), columns = years
by_loc = raw.groupby(["Location", "Year"])[keep].mean().T
by_loc = by_loc.stack("Location").swaplevel().sort_index()   # rows: Location -> element, cols: Year

# attach the coordinates (mean per location)
coords = raw.groupby("Location")[["Latitude", "Longitude"]].mean().round(4)

pd.set_option("display.width", 300)
pd.set_option("display.max_columns", 30)
print(coords)
by_loc.round(3).head(32)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

raw = pd.read_csv("dataset/delhi_20yr.csv")

keep = ["TDS", "TH", "F", "NO3", "K",
        "As", "Pb", "Cd", "Cr", "Ni", "Cu", "Zn", "Fe", "Mn", "pH"]

# one row per (Location, Year): mean of each element, coordinates, unsafe share
agg_dict = {c: "mean" for c in keep}
agg_dict.update({"Latitude": "mean", "Longitude": "mean", "Is_Unsafe": "mean"})
loc_year = raw.groupby(["Location", "Year"]).agg(agg_dict).reset_index()

# how many samples sit behind each row
counts = raw.groupby(["Location", "Year"]).size()
print("Rows:", loc_year.shape[0], "| samples per row: min", counts.min(),
      "max", counts.max(), "mean", round(counts.mean(), 1))

# Spearman on the numeric columns (Location is text, so it is left out)
corr = loc_year.drop(columns="Location").corr(method="spearman")

pd.set_option("display.width", 300)
pd.set_option("display.max_columns", 30)
print(corr.round(2))

others = ["Year", "Is_Unsafe"]
print("\nCorrelation with Year:")
print(corr["Year"].drop(others).sort_values(ascending=False).round(3))
print("\nCorrelation with Is_Unsafe (unsafe share):")
print(corr["Is_Unsafe"].drop(others).sort_values(ascending=False).round(3))

plt.figure(figsize=(13, 10))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", vmin=-1, vmax=1,
            linewidths=0.4, annot_kws={"size": 8})
plt.title("Spearman correlation: Location-Year averages")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

raw = pd.read_csv("dataset/delhi_20yr.csv")

keep = ["TDS", "TH", "F", "NO3", "K",
        "As", "Pb", "Cd", "Cr", "Ni", "Cu", "Zn", "Fe", "Mn", "pH"]

# elements as rows, years as columns, all locations combined
year_table = raw.groupby("Year")[keep].mean().T
year_table.loc["Unsafe_share"] = raw.groupby("Year")["Is_Unsafe"].mean()

# rescale each row to 0-1 for color only
scaled = year_table.sub(year_table.min(axis=1), axis=0).div(
    year_table.max(axis=1) - year_table.min(axis=1), axis=0)

plt.figure(figsize=(16, 7))
sns.heatmap(scaled, annot=year_table.round(2), fmt="", cmap="YlOrRd",
            linewidths=0.4, annot_kws={"size": 7},
            cbar_kws={"label": "Row-scaled (0 = lowest year, 1 = highest year)"})
plt.title("Yearly mean of each element, all locations combined")
plt.xlabel("Year")
plt.ylabel("")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

raw = pd.read_csv("dataset/delhi_20yr.csv")

counts = raw["Is_Unsafe"].value_counts().sort_index()   # 0 = Safe, 1 = Unsafe
labels = ["Safe", "Unsafe"]
print(counts.to_dict())

plt.figure(figsize=(6, 6))
plt.pie(counts.values, labels=labels, explode=[0.12, 0],
        colors=["#2e8b57", "#d9534f"], startangle=90,
        autopct=lambda p: f"{p:.1f}%\n({round(p*counts.sum()/100)})",
        pctdistance=0.75)
plt.title("Safe vs Unsafe samples (all years, all locations)")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

raw = pd.read_csv("dataset/delhi_20yr.csv")

keep = ["TDS", "TH", "F", "NO3", "K",
        "As", "Pb", "Cd", "Cr", "Ni", "Cu", "Zn", "Fe", "Mn", "pH"]

# 1) outlier share per element (IQR rule)
q1, q3 = raw[keep].quantile(0.25), raw[keep].quantile(0.75)
iqr = q3 - q1
is_out = (raw[keep] < q1 - 1.5 * iqr) | (raw[keep] > q3 + 1.5 * iqr)
print((is_out.mean() * 100).round(1).sort_values(ascending=False))

# 2) one boxplot panel per element, each with its own y-axis
fig, axes = plt.subplots(3, 5, figsize=(16, 9))
for ax, col in zip(axes.ravel(), keep):
    ax.boxplot(raw[col], flierprops=dict(marker="o", markersize=3, alpha=0.5))
    ax.set_title(col)
    ax.set_xticks([])
plt.suptitle("Boxplots of the 15 elements (dots = IQR outliers)")
plt.tight_layout()
plt.show()
```
